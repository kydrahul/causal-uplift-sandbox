# 01 — Exploratory Data Analysis
Understand the MovieLens 1M dataset before building features.

```python
import sys
sys.path.insert(0, "..")
from src.data_loader import load_movielens_1m, load_merged
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid", font_scale=1.1)
```

```python
ratings, users, movies = load_movielens_1m()
print(ratings.shape, users.shape, movies.shape)
ratings.head()
```

## Rating distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
ratings["rating"].value_counts().sort_index().plot(kind="bar", ax=axes[0], color="steelblue")
axes[0].set_title("Rating count by score")
axes[0].set_xlabel("Rating")

ratings_per_user = ratings.groupby('user_id').size()
ratings_per_user.hist(bins=50, ax=axes[1], color='tomato', edgecolor='white')
axes[1].set_title("Ratings per user")
axes[1].set_xlabel("# ratings")
plt.tight_layout()
plt.savefig("../results/figures/eda_ratings.png", bbox_inches="tight")
plt.show()
```

## User demographics

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
users["gender"].value_counts().plot(kind="bar", ax=axes[0], color=["salmon", "steelblue"])
axes[0].set_title("Gender distribution")

age_map = {1: '<18', 18: '18-24', 25: '25-34', 35: '35-44', 45: '45-49', 50: '50-55', 56: '56+'}
users["age_label"] = users["age"].map(age_map)
users["age_label"].value_counts().sort_index().plot(kind="bar", ax=axes[1], color="mediumpurple")
axes[1].set_title("Age distribution")

users["occupation"].value_counts().plot(kind="bar", ax=axes[2], color="mediumseagreen")
axes[2].set_title("Occupation distribution (code)")
plt.tight_layout()
plt.savefig("../results/figures/eda_users.png", bbox_inches="tight")
plt.show()
```

## Genre analysis

```python
genres_flat = []
for g in movies["genres"]:
    genres_flat.extend(g.split("|"))
gene_counts = pd.Series(genres_flat).value_counts()

plt.figure(figsize=(12, 5))
sns.barplot(x=gene_counts.values, y=gene_counts.index, palette="viridis")
plt.title("Movie genre frequency")
plt.xlabel("Count")
plt.tight_layout()
plt.savefig("../results/figures/eda_genres.png", bbox_inches="tight")
plt.show()
```

## Temporal activity pattern

```python
merged = load_merged()
merged["month"] = merged["datetime"].dt.to_period("M")
monthly = merged.groupby("month").size()
monthly.plot(figsize=(12, 4), color="steelblue")
plt.title("Monthly rating activity")
plt.xlabel("Month")
plt.ylabel("# ratings")
plt.tight_layout()
plt.savefig("../results/figures/eda_temporal.png", bbox_inches="tight")
plt.show()
```
